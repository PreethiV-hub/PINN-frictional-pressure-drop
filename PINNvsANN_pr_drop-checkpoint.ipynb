```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
os.environ["OMP_NUM_THREADS"] = "1"
import copy
import random
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, balanced_accuracy_score

# ============================================================
# 1. Configuration -- Shared Hyperparameters
# ============================================================
SEED = 42
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
G = 9.81
DATASET_PATH = r"C:\Users\Preethi\OneDrive\Documents\newdata2.xlsx"
OUTPUT_DIR = Path("./annvspinn")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

TARGET_REGIME_COL = "Flow Regime"
PHI_COL = "phi_L^2, Lockhart-Martinelli Multiplier (-)"
DP_COL = "(dP/dz)_TP, Two-Phase Pressure Gradient (Pa/m)"

REGIME_MAP = {"Slug": 0, "Annular": 1, "Stratified": 2}
REGIME_NAMES = {0: "Slug", 1: "Annular", 2: "Stratified"}
REGIME_COLORS = {0: "#1f77b4", 1: "#ff7f0e", 2: "#2ca02c"}
NUM_CLASSES = len(REGIME_MAP)

FEATURE_COLUMNS = [
    "j_L, Liquid Superficial Velocity (m/s)",
    "j_G, Gas Superficial Velocity (m/s)",
    "rho_L, Liquid Density (kg/m3)",
    "rho_G, Gas Density (kg/m3)",
    "D, Pipe Inner Diameter (m)",
    "K, Consistency Index (Pa.s^n)",
    "n, Flow Behaviour Index (-)",
    "theta, Pipe Inclination (deg)"
]
REQUIRED_COLUMNS = FEATURE_COLUMNS + [DP_COL, TARGET_REGIME_COL, PHI_COL]

TEST_SIZE = 0.30
VAL_SIZE_OF_DEV = 0.20
EPOCHS = 600
LEARNING_RATE = 1e-3
PATIENCE = 60

HIDDEN_DIM = 16
WEIGHT_DECAY = 1e-5
LAMBDA_PHI = 0.5
LAMBDA_REGIME = 0.10
DP_LOG_CLAMP = (-1.0, 6.0)
PHI_LOG_CLAMP = (-3.0, 3.0)
SEEDS = [42, 123, 456, 789, 2026]

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
set_seed()

# ============================================================
# 2. Dataset Loading & Preprocessing
# ============================================================
def load_dataset(path):
    df = pd.read_excel(path)
    df.columns = df.columns.astype(str).str.strip()
    missing = [c for c in REQUIRED_COLUMNS if c not in df.columns]
    if missing:
        raise ValueError("Missing columns:\n" + "\n".join(missing))
    numeric_columns = FEATURE_COLUMNS + [DP_COL, PHI_COL]
    for col in numeric_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
    before = len(df)
    df = df.dropna(subset=REQUIRED_COLUMNS).copy()
    positive_columns = [c for c in FEATURE_COLUMNS if c != "theta, Pipe Inclination (deg)"] + [DP_COL, PHI_COL]
    positive_mask = np.ones(len(df), dtype=bool)
    for col in positive_columns:
        positive_mask &= (df[col].to_numpy() > 0)
    df = df.loc[positive_mask].copy()
    df[TARGET_REGIME_COL] = df[TARGET_REGIME_COL].astype(str).str.strip()
    df = df[df[TARGET_REGIME_COL].isin(REGIME_MAP)].copy()
    df["regime_id"] = df[TARGET_REGIME_COL].map(REGIME_MAP).astype(int)
    LI_2013_NAME = "Li, Wong, Skote & Duan (2013) - Chem. Eng. Sci. 102, 121-128"
    if "Source Paper" in df.columns:
        df = df[df["Source Paper"].astype(str).str.strip() != LI_2013_NAME].copy()
    print("\nDataset summary")
    print("-" * 60)
    print(f"Rows before cleaning: {before}")
    print(f"Rows after cleaning : {len(df)}")
    print("\nRegime counts:")
    print(df[TARGET_REGIME_COL].value_counts())
    return df

# ============================================================
# 3. Physics & Feature Transformations
# ============================================================
def reynolds_metzner_reed(rho_L, j_L, D, K, n):
    j_L = torch.clamp(torch.abs(j_L), min=1e-8)
    D = torch.clamp(D, min=1e-8)
    K = torch.clamp(K, min=1e-12)
    n = torch.clamp(n, min=1e-6)
    numerator = rho_L * D.pow(n) * j_L.pow(2.0 - n)
    denominator = (8.0 ** (n - 1.0)) * K * ((3.0 * n + 1.0) / (4.0 * n)).pow(n)
    return numerator / torch.clamp(denominator, min=1e-12)

def liquid_alone_dp_dz(rho_L, j_L, D, K, n):
    Re_MR = reynolds_metzner_reed(rho_L, j_L, D, K, n)
    f_L = 16.0 / torch.clamp(Re_MR, min=1e-8)
    dp_spL = 2.0 * f_L * rho_L * j_L.pow(2) / torch.clamp(D, min=1e-8)
    return dp_spL, Re_MR, f_L

def mixture_properties(rho_L, rho_G, j_L, j_G):
    v_m = j_L + j_G
    alpha = j_G / torch.clamp(v_m, min=1e-8)
    rho_m = alpha * rho_G + (1.0 - alpha) * rho_L
    return v_m, rho_m

def make_raw_features(j_L, j_G, rho_L, rho_G, D, K, n):
    Re_MR = reynolds_metzner_reed(rho_L, j_L, D, K, n)
    Fr_L = j_L / torch.sqrt(G * torch.clamp(D, min=1e-8))
    j_ratio = torch.clamp(j_L / torch.clamp(j_G, min=1e-8), min=1e-8)
    return torch.stack([
        torch.log10(torch.clamp(Re_MR, min=1.0)),
        torch.log10(torch.clamp(Fr_L, min=1e-6)),
        torch.log10(j_ratio),
        torch.log10(torch.clamp(rho_L / torch.clamp(rho_G, min=1e-8), min=1e-8)),
        n
    ], dim=1)

def make_features_from_set(data):
    return make_raw_features(*data[:7])

def fit_feature_normalizer(train_set):
    with torch.no_grad():
        raw_features = make_features_from_set(train_set)
        mean = raw_features.mean(dim=0, keepdim=True)
        std = raw_features.std(dim=0, keepdim=True, unbiased=False).clamp_min(1e-6)
    return mean, std

def dataframe_to_tensors(df):
    tensors = [torch.tensor(df[col].to_numpy(), dtype=torch.float32, device=DEVICE) for col in FEATURE_COLUMNS]
    theta_index = FEATURE_COLUMNS.index("theta, Pipe Inclination (deg)")
    tensors[theta_index] = torch.deg2rad(tensors[theta_index])
    dp_meas = torch.tensor(df[DP_COL].to_numpy(), dtype=torch.float32, device=DEVICE)
    regime_true = torch.tensor(df["regime_id"].to_numpy(), dtype=torch.long, device=DEVICE)
    phi_meas = torch.tensor(df[PHI_COL].to_numpy(), dtype=torch.float32, device=DEVICE)
    return tuple(tensors + [dp_meas, regime_true, phi_meas])

def create_three_way_split(df, seed=42):
    indices = np.arange(len(df))
    labels = df["regime_id"].to_numpy()
    development_idx, test_idx = train_test_split(
        indices, test_size=TEST_SIZE, random_state=seed, stratify=labels
    )
    train_idx, validation_idx = train_test_split(
        development_idx, test_size=VAL_SIZE_OF_DEV, random_state=seed, stratify=labels[development_idx]
    )
    return df.iloc[train_idx].copy(), df.iloc[validation_idx].copy(), df.iloc[test_idx].copy()

def compute_class_weights(regime_true):
    counts = torch.bincount(regime_true, minlength=NUM_CLASSES).float()
    counts = torch.clamp(counts, min=1.0)
    return counts.sum() / (NUM_CLASSES * counts)

# ============================================================
# 4. Model Architectures
# ============================================================
class ANNBaseline(nn.Module):
    def __init__(self, hidden_dim, num_classes, feature_mean, feature_std,
                 phi_bias_init=0.0, dp_bias_init=0.0):
        super().__init__()
        self.register_buffer("feature_mean", feature_mean.detach().clone())
        self.register_buffer("feature_std", feature_std.detach().clone().clamp_min(1e-6))
        self.shared = nn.Sequential(
            nn.Linear(6, hidden_dim), nn.Tanh(),
            nn.Linear(hidden_dim, hidden_dim), nn.Tanh(),
        )
        self.regime_head = nn.Linear(hidden_dim, num_classes)
        self.phi_head = nn.Linear(hidden_dim, 1)
        self.dp_head = nn.Linear(hidden_dim, 1)
        nn.init.zeros_(self.regime_head.bias)
        nn.init.xavier_uniform_(self.phi_head.weight, gain=0.01)
        nn.init.xavier_uniform_(self.dp_head.weight, gain=0.01)
        nn.init.constant_(self.phi_head.bias, phi_bias_init)
        nn.init.constant_(self.dp_head.bias, dp_bias_init)

    def build_features(self, j_L, j_G, rho_L, rho_G, D, K, n, theta):
        raw = make_raw_features(j_L, j_G, rho_L, rho_G, D, K, n)
        norm = (raw - self.feature_mean) / self.feature_std
        return torch.cat([norm, theta.unsqueeze(-1)], dim=1)

    def forward(self, j_L, j_G, rho_L, rho_G, D, K, n, theta):
        features = self.build_features(j_L, j_G, rho_L, rho_G, D, K, n, theta)
        h = self.shared(features)
        phi_pred = 10 ** torch.clamp(self.phi_head(h).squeeze(-1), *PHI_LOG_CLAMP)
        dp_pred = 10 ** torch.clamp(self.dp_head(h).squeeze(-1), *DP_LOG_CLAMP)
        regime_logits = self.regime_head(h)
        return phi_pred, dp_pred, regime_logits

class RegimeConditionedPINN(nn.Module):
    def __init__(self, hidden_dim, num_classes, feature_mean, feature_std):
        super().__init__()
        self.register_buffer("feature_mean", feature_mean.detach().clone())
        self.register_buffer("feature_std", feature_std.detach().clone().clamp_min(1e-6))
        self.classifier = nn.Sequential(
            nn.Linear(5, hidden_dim), nn.Tanh(),
            nn.Linear(hidden_dim, num_classes)
        )
        self.shared = nn.Sequential(nn.Linear(5, hidden_dim), nn.Tanh())
        self.regime_heads = nn.ModuleList([
            nn.Sequential(nn.Linear(hidden_dim, hidden_dim), nn.Tanh(), nn.Linear(hidden_dim, 1))
            for _ in range(num_classes)
        ])
        for head in self.regime_heads:
            nn.init.zeros_(head[-1].bias)

    def build_features(self, j_L, j_G, rho_L, rho_G, D, K, n):
        raw = make_raw_features(j_L, j_G, rho_L, rho_G, D, K, n)
        return (raw - self.feature_mean) / self.feature_std

    def forward(self, j_L, j_G, rho_L, rho_G, D, K, n, regime_true=None, use_true_regime=False):
        features = self.build_features(j_L, j_G, rho_L, rho_G, D, K, n)
        regime_logits = self.classifier(features)
        regime_probs = torch.softmax(regime_logits, dim=1)
        shared_features = self.shared(features)
        head_outputs = torch.cat([head(shared_features) for head in self.regime_heads], dim=1)
        if use_true_regime:
            regime_input = F.one_hot(regime_true, num_classes=NUM_CLASSES).float()
        else:
            regime_input = regime_probs
        log_phi_pred = (head_outputs * regime_input).sum(dim=1)
        phi_pred = 10 ** torch.clamp(log_phi_pred, min=-3.0, max=3.0)
        return phi_pred, regime_logits, regime_probs, head_outputs

    def get_physics_dp(self, phi_pred, j_L, j_G, rho_L, rho_G, D, K, n, theta):
        dp_spL, Re_MR, f_L = liquid_alone_dp_dz(rho_L, j_L, D, K, n)
        v_m, rho_m = mixture_properties(rho_L, rho_G, j_L, j_G)
        dp_gravity = rho_m * G * torch.sin(theta)
        dp_pred = phi_pred * dp_spL + dp_gravity
        return dp_pred

# ============================================================
# 5. Training Routines
# ============================================================
def train_ann(train_set, validation_set, seed):
    class_weights = compute_class_weights(train_set[9]).to(DEVICE)
    criterion_regime = nn.CrossEntropyLoss(weight=class_weights)
    feature_mean, feature_std = fit_feature_normalizer(train_set)
    phi_bias_init = torch.log10(train_set[10].clamp(min=1e-8)).mean().item()
    dp_bias_init = torch.log10(train_set[8].clamp(min=1e-3)).mean().item()
    model = ANNBaseline(HIDDEN_DIM, NUM_CLASSES, feature_mean, feature_std,
                         phi_bias_init, dp_bias_init).to(DEVICE)
    decay_params = [p for n, p in model.named_parameters() if not n.endswith("bias")]
    no_decay_params = [p for n, p in model.named_parameters() if n.endswith("bias")]
    optimizer = optim.AdamW([
        {"params": decay_params, "weight_decay": WEIGHT_DECAY},
        {"params": no_decay_params, "weight_decay": 0.0}
    ], lr=LEARNING_RATE)

    def compute_loss(data):
        j_L, j_G, rho_L, rho_G, D, K, n, theta, dp_meas, regime_true, phi_meas = data
        phi_pred, dp_pred, regime_logits = model(j_L, j_G, rho_L, rho_G, D, K, n, theta)
        log_dp_sq = (torch.log10(torch.clamp(dp_pred, min=1e-3)) - torch.log10(torch.clamp(dp_meas, min=1e-3))).pow(2)
        log_phi_sq = (torch.log10(torch.clamp(phi_pred, min=1e-8)) - torch.log10(torch.clamp(phi_meas, min=1e-8))).pow(2)
        dp_losses, phi_losses = [], []
        for r in range(NUM_CLASSES):
            mask = regime_true == r
            if mask.sum() > 0:
                dp_losses.append(log_dp_sq[mask].mean())
                phi_losses.append(log_phi_sq[mask].mean())
        loss_dp = torch.stack(dp_losses).mean()
        loss_phi = torch.stack(phi_losses).mean()
        loss_regime = criterion_regime(regime_logits, regime_true)
        return loss_dp + LAMBDA_PHI * loss_phi + LAMBDA_REGIME * loss_regime

    best_val, best_state, epochs_no_improve = float("inf"), None, 0
    history = {"train_loss": [], "val_loss": []}

    for epoch in range(1, EPOCHS + 1):
        model.train()
        optimizer.zero_grad(set_to_none=True)
        loss = compute_loss(train_set)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()

        model.eval()
        with torch.inference_mode():
            val_loss = compute_loss(validation_set).item()

        history["train_loss"].append(loss.item())
        history["val_loss"].append(val_loss)

        if val_loss < best_val - 1e-5:
            best_val, best_state, epochs_no_improve = val_loss, copy.deepcopy(model.state_dict()), 0
        else:
            epochs_no_improve += 1

        if epochs_no_improve >= PATIENCE:
            remaining = EPOCHS - epoch
            history["train_loss"].extend([history["train_loss"][-1]] * remaining)
            history["val_loss"].extend([history["val_loss"][-1]] * remaining)
            break

    if best_state is not None:
        model.load_state_dict(best_state)
    return model, history

def train_pinn(train_set, validation_set, seed):
    class_weights = compute_class_weights(train_set[9]).to(DEVICE)
    criterion_regime = nn.CrossEntropyLoss(weight=class_weights)
    feature_mean, feature_std = fit_feature_normalizer(train_set)
    model = RegimeConditionedPINN(HIDDEN_DIM, NUM_CLASSES, feature_mean, feature_std).to(DEVICE)
    optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)

    def compute_loss(data, use_true_regime):
        j_L, j_G, rho_L, rho_G, D, K, n, theta, dp_meas, regime_true, phi_meas = data
        phi_pred, regime_logits, regime_probs, head_outputs = model(
            j_L, j_G, rho_L, rho_G, D, K, n, regime_true=regime_true, use_true_regime=use_true_regime
        )
        dp_pred = model.get_physics_dp(phi_pred, j_L, j_G, rho_L, rho_G, D, K, n, theta)
        log_dp_sq = (torch.log10(torch.clamp(dp_pred, min=1e-3)) - torch.log10(torch.clamp(dp_meas, min=1e-3))).pow(2)
        log_phi_sq = (torch.log10(torch.clamp(phi_pred, min=1e-8)) - torch.log10(torch.clamp(phi_meas, min=1e-8))).pow(2)
        dp_losses, phi_losses = [], []
        for r in range(NUM_CLASSES):
            mask = regime_true == r
            if mask.sum() > 0:
                dp_losses.append(log_dp_sq[mask].mean())
                phi_losses.append(log_phi_sq[mask].mean())
        loss_dp = torch.stack(dp_losses).mean()
        loss_phi = torch.stack(phi_losses).mean()
        loss_regime = criterion_regime(regime_logits, regime_true)
        return loss_dp + LAMBDA_PHI * loss_phi + LAMBDA_REGIME * loss_regime

    best_val, best_state, epochs_no_improve = float("inf"), None, 0
    history = {"train_loss": [], "val_loss": []}

    for epoch in range(1, EPOCHS + 1):
        model.train()
        optimizer.zero_grad(set_to_none=True)
        loss = compute_loss(train_set, use_true_regime=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()

        model.eval()
        with torch.inference_mode():
            val_loss = compute_loss(validation_set, use_true_regime=False).item()

        history["train_loss"].append(loss.item())
        history["val_loss"].append(val_loss)

        if val_loss < best_val - 1e-5:
            best_val, best_state, epochs_no_improve = val_loss, copy.deepcopy(model.state_dict()), 0
        else:
            epochs_no_improve += 1

        if epochs_no_improve >= PATIENCE:
            remaining = EPOCHS - epoch
            history["train_loss"].extend([history["train_loss"][-1]] * remaining)
            history["val_loss"].extend([history["val_loss"][-1]] * remaining)
            break

    if best_state is not None:
        model.load_state_dict(best_state)
    return model, history

# ============================================================
# 6. Evaluation Functions
# ============================================================
def log10_rmse(y_true, y_pred):
    y_true = torch.clamp(y_true, min=1e-8)
    y_pred = torch.clamp(y_pred, min=1e-8)
    log_err = torch.log10(y_pred) - torch.log10(y_true)
    return torch.sqrt(torch.mean(log_err.pow(2))).item()

def median_ape(y_true, y_pred):
    y_true = torch.clamp(y_true, min=1e-8)
    y_pred = torch.clamp(y_pred, min=1e-8)
    ape = torch.abs((y_pred - y_true) / y_true) * 100.0
    return torch.median(ape).item()

def evaluate_ann(model, test_set):
    model.eval()
    with torch.inference_mode():
        j_L, j_G, rho_L, rho_G, D, K, n, theta = test_set[:8]
        phi_pred, dp_pred, regime_logits = model(j_L, j_G, rho_L, rho_G, D, K, n, theta)
    regime_pred = regime_logits.argmax(dim=1)
    return {
        "dp_true": test_set[8].cpu(), "dp_pred": dp_pred.detach().cpu(),
        "phi_true": test_set[10].cpu(), "phi_pred": phi_pred.detach().cpu(),
        "regime_true": test_set[9].cpu(), "regime_pred": regime_pred.detach().cpu()
    }

def evaluate_pinn(model, test_set):
    model.eval()
    with torch.inference_mode():
        j_L, j_G, rho_L, rho_G, D, K, n, theta = test_set[:8]
        phi_pred, regime_logits, regime_probs, head_outputs = model(
            j_L, j_G, rho_L, rho_G, D, K, n, regime_true=None, use_true_regime=False
        )
        dp_pred = model.get_physics_dp(phi_pred, j_L, j_G, rho_L, rho_G, D, K, n, theta)
    regime_pred = regime_logits.argmax(dim=1)
    return {
        "dp_true": test_set[8].cpu(), "dp_pred": dp_pred.detach().cpu(),
        "phi_true": test_set[10].cpu(), "phi_pred": phi_pred.detach().cpu(),
        "regime_true": test_set[9].cpu(), "regime_pred": regime_pred.detach().cpu()
    }

def summarize_run(results):
    acc = accuracy_score(results["regime_true"], results["regime_pred"]) * 100.0
    bal_acc = balanced_accuracy_score(results["regime_true"], results["regime_pred"]) * 100.0
    return {
        "classification_accuracy": acc,
        "balanced_accuracy": bal_acc,
        "dp_log10_rmse": log10_rmse(results["dp_true"], results["dp_pred"]),
        "dp_median_ape": median_ape(results["dp_true"], results["dp_pred"]),
        "phi_log10_rmse": log10_rmse(results["phi_true"], results["phi_pred"]),
        "phi_median_ape": median_ape(results["phi_true"], results["phi_pred"]),
    }

def per_regime_metrics(results):
    rows = []
    for regime_id in range(NUM_CLASSES):
        mask = results["regime_true"] == regime_id
        if mask.sum() == 0:
            continue
        rows.append({
            "Regime": REGIME_NAMES[regime_id],
            "N": int(mask.sum().item()),
            "dp_log10_rmse": log10_rmse(results["dp_true"][mask], results["dp_pred"][mask]),
            "dp_median_ape": median_ape(results["dp_true"][mask], results["dp_pred"][mask]),
            "phi_log10_rmse": log10_rmse(results["phi_true"][mask], results["phi_pred"][mask]),
            "phi_median_ape": median_ape(results["phi_true"][mask], results["phi_pred"][mask]),
        })
    return pd.DataFrame(rows)

def pool_results(results_list):
    keys = results_list[0].keys()
    return {k: torch.cat([r[k] for r in results_list], dim=0) for k in keys}

def calc_within_band_percentage(y_true, y_pred, threshold=0.20):
    rel_err = np.abs((y_pred - y_true) / y_true)
    within = np.mean(rel_err <= threshold) * 100.0
    return within

# Helper function to plot a single parity subplot onto any provided Matplotlib axis
def render_single_parity_plot(ax, y_true, y_pred, regimes, xlabel, ylabel, title_prefix, min_default, max_default):
    overall_within = calc_within_band_percentage(y_true, y_pred, 0.20)

    for r_id, r_name in REGIME_NAMES.items():
        mask = (regimes == r_id)
        if np.sum(mask) > 0:
            r_within = calc_within_band_percentage(y_true[mask], y_pred[mask], 0.20)
            ax.scatter(y_true[mask], y_pred[mask], c=REGIME_COLORS[r_id], 
                       label=f"{r_name} ({r_within:.1f}% in ±20%)", alpha=0.6, edgecolors="none", s=25)

    min_val = min(y_true.min(), y_pred.min(), min_default)
    max_val = max(y_true.max(), y_pred.max(), max_default)

    ax.plot([min_val, max_val], [min_val, max_val], 'k--', lw=1.5, label="Ideal")
    ax.plot([min_val, max_val], [min_val * 1.2, max_val * 1.2], 'k:', alpha=0.7, label="±20% Band")
    ax.plot([min_val, max_val], [min_val * 0.8, max_val * 0.8], 'k:', alpha=0.7)
    ax.fill_between([min_val, max_val], [min_val * 0.8, max_val * 0.8], [min_val * 1.2, max_val * 1.2], color="gray", alpha=0.1)

    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_xlim([min_val, max_val])
    ax.set_ylim([min_val, max_val])
    ax.set_xlabel(xlabel, fontsize=11)
    ax.set_ylabel(ylabel, fontsize=11)
    ax.set_title(f"{title_prefix}\nOverall in ±20%: {overall_within:.1f}%", fontsize=12, fontweight="bold")
    ax.grid(True, which="both", linestyle="--", alpha=0.3)
    ax.legend(loc="upper left", fontsize=9)

# ============================================================
# 7. Plotting Functions (Grid + Individual Exports)
# ============================================================
def plot_parity(ann_pooled, pinn_pooled, combined_path, output_dir):
    # 1. Combined 2x2 Grid Image
    fig, axes = plt.subplots(2, 2, figsize=(14, 12))

    # Pressure Gradient - ANN
    render_single_parity_plot(
        axes[0, 0], ann_pooled["dp_true"].numpy(), ann_pooled["dp_pred"].numpy(), ann_pooled["regime_true"].numpy(),
        r"Measured $(dP/dz)_{TP}$ (Pa/m)", r"Predicted $(dP/dz)_{TP}$ (Pa/m)", "Pressure Gradient (ANN)", 1e-1, 1e5
    )
    # Pressure Gradient - PINN
    render_single_parity_plot(
        axes[0, 1], pinn_pooled["dp_true"].numpy(), pinn_pooled["dp_pred"].numpy(), pinn_pooled["regime_true"].numpy(),
        r"Measured $(dP/dz)_{TP}$ (Pa/m)", r"Predicted $(dP/dz)_{TP}$ (Pa/m)", "Pressure Gradient (PINN)", 1e-1, 1e5
    )
    # Multiplier - ANN
    render_single_parity_plot(
        axes[1, 0], ann_pooled["phi_true"].numpy(), ann_pooled["phi_pred"].numpy(), ann_pooled["regime_true"].numpy(),
        r"Measured $\phi_L^2$ (-)", r"Predicted $\phi_L^2$ (-)", r"Multiplier $\phi_L^2$ (ANN)", 1e-2, 1e3
    )
    # Multiplier - PINN
    render_single_parity_plot(
        axes[1, 1], pinn_pooled["phi_true"].numpy(), pinn_pooled["phi_pred"].numpy(), pinn_pooled["regime_true"].numpy(),
        r"Measured $\phi_L^2$ (-)", r"Predicted $\phi_L^2$ (-)", r"Multiplier $\phi_L^2$ (PINN)", 1e-2, 1e3
    )

    fig.tight_layout()
    fig.savefig(combined_path, dpi=200)
    plt.close(fig)
    print(f"Saved combined 2x2 parity plots to: {combined_path}")

    # 2. Individual Single Images
    configs = [
        ("parity_dp_ann.png", ann_pooled["dp_true"].numpy(), ann_pooled["dp_pred"].numpy(), ann_pooled["regime_true"].numpy(),
         r"Measured $(dP/dz)_{TP}$ (Pa/m)", r"Predicted $(dP/dz)_{TP}$ (Pa/m)", "Pressure Gradient (ANN)", 1e-1, 1e5),
        ("parity_dp_pinn.png", pinn_pooled["dp_true"].numpy(), pinn_pooled["dp_pred"].numpy(), pinn_pooled["regime_true"].numpy(),
         r"Measured $(dP/dz)_{TP}$ (Pa/m)", r"Predicted $(dP/dz)_{TP}$ (Pa/m)", "Pressure Gradient (PINN)", 1e-1, 1e5),
        ("parity_phi_ann.png", ann_pooled["phi_true"].numpy(), ann_pooled["phi_pred"].numpy(), ann_pooled["regime_true"].numpy(),
         r"Measured $\phi_L^2$ (-)", r"Predicted $\phi_L^2$ (-)", r"Multiplier $\phi_L^2$ (ANN)", 1e-2, 1e3),
        ("parity_phi_pinn.png", pinn_pooled["phi_true"].numpy(), pinn_pooled["phi_pred"].numpy(), pinn_pooled["regime_true"].numpy(),
         r"Measured $\phi_L^2$ (-)", r"Predicted $\phi_L^2$ (-)", r"Multiplier $\phi_L^2$ (PINN)", 1e-2, 1e3),
    ]

    for fname, y_t, y_p, reg, x_lbl, y_lbl, title, min_d, max_d in configs:
        fig_ind, ax_ind = plt.subplots(figsize=(7, 6))
        render_single_parity_plot(ax_ind, y_t, y_p, reg, x_lbl, y_lbl, title, min_d, max_d)
        fig_ind.tight_layout()
        ind_path = output_dir / fname
        fig_ind.savefig(ind_path, dpi=200)
        plt.close(fig_ind)
        print(f"Saved individual parity plot to: {ind_path}")

def plot_loss_curves(ann_histories, pinn_histories, path):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    epochs_arr = np.arange(1, EPOCHS + 1)

    for idx, (label, histories, color) in enumerate([("ANN", ann_histories, "#4C72B0"), ("PINN", pinn_histories, "#DD8452")]):
        ax = axes[idx]
        train_arr = np.array([h["train_loss"] for h in histories])
        val_arr = np.array([h["val_loss"] for h in histories])

        train_mean, train_std = train_arr.mean(axis=0), train_arr.std(axis=0)
        val_mean, val_std = val_arr.mean(axis=0), val_arr.std(axis=0)

        ax.plot(epochs_arr, train_mean, label="Train Loss (Mean)", color=color, linestyle="-", lw=2)
        ax.fill_between(epochs_arr, train_mean - train_std, train_mean + train_std, color=color, alpha=0.15)

        ax.plot(epochs_arr, val_mean, label="Validation Loss (Mean)", color=color, linestyle="--", lw=2)
        ax.fill_between(epochs_arr, val_mean - val_std, val_mean + val_std, color=color, alpha=0.3)

        ax.set_yscale("log")
        ax.set_xlabel("Epoch", fontsize=11)
        ax.set_ylabel("Loss (Log Scale)", fontsize=11)
        ax.set_title(f"Average Training & Validation Loss across 5 Seeds ({label})", fontsize=12, fontweight="bold")
        ax.grid(True, which="both", linestyle="--", alpha=0.4)
        ax.legend(loc="upper right")

    fig.tight_layout()
    fig.savefig(path, dpi=200)
    plt.close(fig)
    print(f"Saved loss curves to: {path}")

def plot_headline_comparison(ann_summary_df, pinn_summary_df, path):
    metrics = [
        ("dp_log10_rmse", "dP/dz Log10 RMSE"),
        ("dp_median_ape", "dP/dz Median APE (%)"),
        ("phi_log10_rmse", r"$\phi_L^2$ Log10 RMSE"),
        ("classification_accuracy", "Classification Accuracy (%)"),
    ]
    fig, axes = plt.subplots(1, 4, figsize=(18, 4.5))
    for ax, (col, label) in zip(axes, metrics):
        ann_mean, ann_std = ann_summary_df[col].mean(), ann_summary_df[col].std()
        pinn_mean, pinn_std = pinn_summary_df[col].mean(), pinn_summary_df[col].std()
        bars = ax.bar(
            ["ANN", "PINN"], [ann_mean, pinn_mean],
            yerr=[ann_std, pinn_std], capsize=6,
            color=["#4C72B0", "#DD8452"], alpha=0.85
        )
        for bar, val in zip(bars, [ann_mean, pinn_mean]):
            ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(),
                    f"{val:.3f}", ha="center", va="bottom", fontsize=9)
        ax.set_title(label, fontsize=10)
        ax.grid(True, axis="y", linestyle="--", alpha=0.35)
    fig.suptitle("Headline Metrics: ANN vs. PINN (mean ± std over 5 seeds)", fontsize=13)
    fig.tight_layout(rect=[0, 0, 1, 0.94])
    fig.savefig(path, dpi=160)
    plt.close(fig)

def plot_per_regime_comparison(ann_regime_dfs, pinn_regime_dfs, path):
    ann_all = pd.concat(ann_regime_dfs)
    pinn_all = pd.concat(pinn_regime_dfs)
    ann_agg = ann_all.groupby("Regime").agg(
        dp_mean=("dp_log10_rmse", "mean"), dp_std=("dp_log10_rmse", "std"),
        phi_mean=("phi_log10_rmse", "mean"), phi_std=("phi_log10_rmse", "std"),
    ).reindex(["Slug", "Annular", "Stratified"])
    pinn_agg = pinn_all.groupby("Regime").agg(
        dp_mean=("dp_log10_rmse", "mean"), dp_std=("dp_log10_rmse", "std"),
        phi_mean=("phi_log10_rmse", "mean"), phi_std=("phi_log10_rmse", "std"),
    ).reindex(["Slug", "Annular", "Stratified"])
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    x = np.arange(3)
    width = 0.35
    axes[0].bar(x - width / 2, ann_agg["dp_mean"], width, yerr=ann_agg["dp_std"],
                label="ANN", color="#4C72B0", capsize=5, alpha=0.85)
    axes[0].bar(x + width / 2, pinn_agg["dp_mean"], width, yerr=pinn_agg["dp_std"],
                label="PINN", color="#DD8452", capsize=5, alpha=0.85)
    axes[0].set_xticks(x)
    axes[0].set_xticklabels(["Slug", "Annular", "Stratified"])
    axes[0].set_ylabel("dP/dz Log10 RMSE")
    axes[0].set_title("Pressure-Gradient Error by Regime")
    axes[0].legend()
    axes[0].grid(True, axis="y", linestyle="--", alpha=0.35)
    axes[1].bar(x - width / 2, ann_agg["phi_mean"], width, yerr=ann_agg["phi_std"],
                label="ANN", color="#4C72B0", capsize=5, alpha=0.85)
    axes[1].bar(x + width / 2, pinn_agg["phi_mean"], width, yerr=pinn_agg["phi_std"],
                label="PINN", color="#DD8452", capsize=5, alpha=0.85)
    axes[1].set_xticks(x)
    axes[1].set_xticklabels(["Slug", "Annular", "Stratified"])
    axes[1].set_ylabel(r"$\phi_L^2$ Log10 RMSE")
    axes[1].set_title(r"$\phi_L^2$ Error by Regime")
    axes[1].legend()
    axes[1].grid(True, axis="y", linestyle="--", alpha=0.35)
    fig.suptitle("Per-Regime Comparison: ANN vs. PINN (mean ± std over 5 seeds)", fontsize=13)
    fig.tight_layout(rect=[0, 0, 1, 0.93])
    fig.savefig(path, dpi=160)
    plt.close(fig)

def plot_residual_histograms(ann_pooled, pinn_pooled, path):
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    for label, results, color in [("ANN", ann_pooled, "#4C72B0"), ("PINN", pinn_pooled, "#DD8452")]:
        dp_true = torch.clamp(results["dp_true"], min=1e-8)
        dp_pred = torch.clamp(results["dp_pred"], min=1e-8)
        log_resid_dp = (torch.log10(dp_pred) - torch.log10(dp_true)).numpy()
        axes[0].hist(log_resid_dp, bins=30, alpha=0.55, label=label, color=color, density=True)
        phi_true = torch.clamp(results["phi_true"], min=1e-8)
        phi_pred = torch.clamp(results["phi_pred"], min=1e-8)
        log_resid_phi = (torch.log10(phi_pred) - torch.log10(phi_true)).numpy()
        axes[1].hist(log_resid_phi, bins=30, alpha=0.55, label=label, color=color, density=True)
    axes[0].axvline(0, color="black", linestyle="--", linewidth=1)
    axes[0].set_xlabel(r"$\log_{10}(dP/dz_{pred} / dP/dz_{true})$")
    axes[0].set_ylabel("Density")
    axes[0].set_title("Pressure-Gradient Residuals (pooled, 5 seeds)")
    axes[0].legend()
    axes[0].grid(True, linestyle="--", alpha=0.35)
    axes[1].axvline(0, color="black", linestyle="--", linewidth=1)
    axes[1].set_xlabel(r"$\log_{10}(\phi_{pred} / \phi_{true})$")
    axes[1].set_ylabel("Density")
    axes[1].set_title(r"$\phi_L^2$ Residuals (pooled, 5 seeds)")
    axes[1].legend()
    axes[1].grid(True, linestyle="--", alpha=0.35)
    fig.suptitle("Residual Distributions: ANN vs. PINN", fontsize=13)
    fig.tight_layout(rect=[0, 0, 1, 0.93])
    fig.savefig(path, dpi=160)
    plt.close(fig)

# ============================================================
# 8. Tables
# ============================================================
def build_overall_performance_table(ann_summary_df, pinn_summary_df):
    metric_specs = [
        ("classification_accuracy", "Classification accuracy", "higher", "%"),
        ("balanced_accuracy", "Balanced accuracy", "higher", "%"),
        ("dp_log10_rmse", "Pressure log-RMSE", "lower", ""),
        ("dp_median_ape", "Pressure median APE", "lower", "%"),
        ("phi_log10_rmse", "phi_L^2 log-RMSE", "lower", ""),
        ("phi_median_ape", "phi_L^2 median APE", "lower", "%"),
    ]
    rows = []
    for col, label, direction, unit in metric_specs:
        pinn_mean, pinn_std = pinn_summary_df[col].mean(), pinn_summary_df[col].std()
        ann_mean, ann_std = ann_summary_df[col].mean(), ann_summary_df[col].std()
        better = "PINN" if (pinn_mean < ann_mean if direction == "lower" else pinn_mean > ann_mean) else "ANN"
        rows.append({
            "Metric": label,
            "PINN mean +- SD": f"{pinn_mean:.4f} +- {pinn_std:.4f}{unit}",
            "ANN mean +- SD": f"{ann_mean:.4f} +- {ann_std:.4f}{unit}",
            "Better model": better,
        })
    return pd.DataFrame(rows)

def build_regime_performance_table(ann_regime_dfs, pinn_regime_dfs):
    ann_all = pd.concat(ann_regime_dfs)
    pinn_all = pd.concat(pinn_regime_dfs)
    def agg(df, cols):
        out = df.groupby("Regime").agg({c: ["mean", "std"] for c in cols})
        out.columns = ["_".join(c) for c in out.columns]
        return out.reindex(["Slug", "Annular", "Stratified"])
    ann_agg = agg(ann_all, ["phi_log10_rmse", "phi_median_ape", "dp_log10_rmse", "dp_median_ape"])
    pinn_agg = agg(pinn_all, ["phi_log10_rmse", "phi_median_ape", "dp_log10_rmse", "dp_median_ape"])
    phi_rows, dp_rows = [], []
    for regime in ["Slug", "Annular", "Stratified"]:
        phi_rows.append({
            "Flow regime": regime,
            "PINN phi_L^2 log-RMSE": f"{pinn_agg.loc[regime, 'phi_log10_rmse_mean']:.4f} +- {pinn_agg.loc[regime, 'phi_log10_rmse_std']:.4f}",
            "ANN phi_L^2 log-RMSE": f"{ann_agg.loc[regime, 'phi_log10_rmse_mean']:.4f} +- {ann_agg.loc[regime, 'phi_log10_rmse_std']:.4f}",
            "PINN phi_L^2 median APE": f"{pinn_agg.loc[regime, 'phi_median_ape_mean']:.4f}% +- {pinn_agg.loc[regime, 'phi_median_ape_std']:.4f}%",
            "ANN phi_L^2 median APE": f"{ann_agg.loc[regime, 'phi_median_ape_mean']:.4f}% +- {ann_agg.loc[regime, 'phi_median_ape_std']:.4f}%",
        })
        dp_rows.append({
            "Flow regime": regime,
            "PINN pressure log-RMSE": f"{pinn_agg.loc[regime, 'dp_log10_rmse_mean']:.4f} +- {pinn_agg.loc[regime, 'dp_log10_rmse_std']:.4f}",
            "ANN pressure log-RMSE": f"{ann_agg.loc[regime, 'dp_log10_rmse_mean']:.4f} +- {ann_agg.loc[regime, 'dp_log10_rmse_std']:.4f}",
            "PINN pressure median APE": f"{pinn_agg.loc[regime, 'dp_median_ape_mean']:.4f}% +- {pinn_agg.loc[regime, 'dp_median_ape_std']:.4f}%",
            "ANN pressure median APE": f"{ann_agg.loc[regime, 'dp_median_ape_mean']:.4f}% +- {ann_agg.loc[regime, 'dp_median_ape_std']:.4f}%",
        })
    return pd.DataFrame(phi_rows), pd.DataFrame(dp_rows)

# ============================================================
# 9. Main Execution
# ============================================================
if __name__ == "__main__":
    print(f"Running on device: {DEVICE}")
    df = load_dataset(DATASET_PATH)

    ann_summaries, pinn_summaries = [], []
    ann_regime_dfs, pinn_regime_dfs = [], []
    ann_results_list, pinn_results_list = [], []
    ann_histories, pinn_histories = [], []

    for seed in SEEDS:
        print(f"\n--- Seed {seed} ---")
        set_seed(seed)
        train_df, val_df, test_df = create_three_way_split(df, seed=seed)
        train_set = dataframe_to_tensors(train_df)
        val_set = dataframe_to_tensors(val_df)
        test_set = dataframe_to_tensors(test_df)

        set_seed(seed)
        ann_model, ann_hist = train_ann(train_set, val_set, seed)
        ann_results = evaluate_ann(ann_model, test_set)
        ann_summaries.append(summarize_run(ann_results))
        ann_regime_dfs.append(per_regime_metrics(ann_results))
        ann_results_list.append(ann_results)
        ann_histories.append(ann_hist)

        set_seed(seed)
        pinn_model, pinn_hist = train_pinn(train_set, val_set, seed)
        pinn_results = evaluate_pinn(pinn_model, test_set)
        pinn_summaries.append(summarize_run(pinn_results))
        pinn_regime_dfs.append(per_regime_metrics(pinn_results))
        pinn_results_list.append(pinn_results)
        pinn_histories.append(pinn_hist)

    ann_summary_df = pd.DataFrame(ann_summaries)
    pinn_summary_df = pd.DataFrame(pinn_summaries)
    ann_pooled = pool_results(ann_results_list)
    pinn_pooled = pool_results(pinn_results_list)

    plot_headline_comparison(ann_summary_df, pinn_summary_df, OUTPUT_DIR / "1_headline_comparison.png")
    plot_per_regime_comparison(ann_regime_dfs, pinn_regime_dfs, OUTPUT_DIR / "2_per_regime_comparison.png")
    plot_residual_histograms(ann_pooled, pinn_pooled, OUTPUT_DIR / "5_residual_histograms.png")
    
    # Generate both 2x2 combined and individual image files
    plot_parity(ann_pooled, pinn_pooled, OUTPUT_DIR / "3_parity_plots.png", OUTPUT_DIR)
    
    plot_loss_curves(ann_histories, pinn_histories, OUTPUT_DIR / "4_loss_curves.png")

    overall_table = build_overall_performance_table(ann_summary_df, pinn_summary_df)
    phi_regime_table, dp_regime_table = build_regime_performance_table(ann_regime_dfs, pinn_regime_dfs)

    overall_table.to_csv(OUTPUT_DIR / "overall_performance_table.csv", index=False)
    phi_regime_table.to_csv(OUTPUT_DIR / "regime_performance_phi_table.csv", index=False)
    dp_regime_table.to_csv(OUTPUT_DIR / "regime_performance_dp_table.csv", index=False)

    print("\nFinished. Saved combined grid image ('3_parity_plots.png') and individual image files ('parity_dp_ann.png', 'parity_dp_pinn.png', 'parity_phi_ann.png', 'parity_phi_pinn.png').")
```
